```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("homeprices.csv")
```

```python
df
```

Now we need to fit the best line here to find the new datas that come in

Now we need to find the best line that fits here that has a low MSE

```python
%matplotlib inline
plt. scatter(df.area,df.price)
```

```python
reg = LinearRegression()
reg.fit(df[['area']],df.price)
```

```python
reg.predict([[3300]]) #predicted this using the linear regression model output:array([628715.75342466])
```

```python
%matplotlib inline
plt.scatter(df.area,df.price)
plt.plot(df.area,reg.predict(df[['area']]),color='blue')
```

To get the m(Slope)

```python
reg.coef_ #this is the coefficient the slope m
```

This is the intercept

```python
reg.intercept_
```
